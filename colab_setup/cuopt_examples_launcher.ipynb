### Usage Guide

**Branch Selection:** Different branches contain different versions or experimental features. Switch as needed.

**Running Notebooks:**
- **Locally:** Run directly if you have required dependencies (see `requirements.txt`)
- **In Colab:** Only notebooks that have a cell asking for the launcher setup code need it. If the notebook has one:
  1. Click **"📋 Show Setup Code"** button for the notebook
  2. Click **"Open in Colab"** to launch the notebook
  3. Copy the setup code and paste it into the cell that asks for it
  4. Run the notebook's cells in order from the top. The setup code clones the repo, installs any requirements, and navigates to the notebook's folder.
